```python
file = open('my_file.txt', 'w+')
file.writelines(['A line\n', 'another line\n'])
# open() assume 'rt' by default if nothing is supplied a.k.a. 'read text'
# file = open('my_file.txt', rt)

# read() - attempts to read from current cursor position
# seek() - move cursor position 
# readline() - one by one
# readlines() - return list of strings for each line
# write() - overwrites all in file with new content 
# writelines() - write list of strings into separate lines
# there is no writeline() like with readline() 
file.seek(0)
print(file.read())
file.close()

# Base modes:
# r  = read (file must exist; starts at beginning; no writing)
# a  = append (creates file if missing; writes at end; no reading)
# w  = write (creates file if missing; truncates existing file immediately)
# x  = exclusive create (creates new file; raises FileExistsError if it exists)

# + = read and write (behavior depends on the base mode)
# r+ = read & write (file must exist; no truncation; writes at current position)
# a+ = read & append (creates file if missing; writes always go to end)
# w+ = read & write (creates file if missing; truncates immediately)
# x+ = read & write (creates new file; raises FileExistsError if it exists)

# t = text mode (default)
# b = binary mode (e.g., 'rb', 'wb')

import os
import json
from pathlib import Path

with open('my_file.txt', 'r+') as file:
    file.write('I replaced all the content!')
    file.seek(0)
    print(file.read())

if os.path.exists('my_file.txt'):
    os.remove('my_file.txt')

# create if doesn't exist
# Path("myfile.txt").touch(exist_ok=True)

# remove directory (you can only remove if empty)
# os.rmdir('myfolder')

# recently added - 10/26
file_path = Path("my_file.json")

# x = exclusive create; throws FileExistsError and avoids race condition if 
# another process tries to create the file as this one runs 
# json modules lets us write python dictionaries to json and vice versa 
if not file_path.exists():
    try:
        with open("my_file.json", "x") as file:
            json.dump({"Test Key":"Test Value"}, file, indent=4)
    except FileExistsError:
        pass

with file_path.open("r") as file:
    data = json.load(file)
    print(data)


# json.dumps(my_dict)
# Converts a dictionary into a JSON string

# json.dump(my_dict, file)
# Writes a dictionary directly into a JSON file

# json.loads(json_string)
# Converts a JSON string into a Python object

# json.load(file)
# Reads a JSON file into a Python object
```
